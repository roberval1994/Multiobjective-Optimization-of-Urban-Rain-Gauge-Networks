# Interpolação por krigagem
Referência Clássica (Teoria):

MATHERON, G. (1963). Principles of geostatistics. Economic Geology, v. 58, n. 8, p. 1246-1266.

Referência Aplicada a Chuvas (WMO):

WORLD METEOROLOGICAL ORGANIZATION (WMO). Guide to Hydrological Practices: Volume I: Hydrology - From Measurement to Hydrological Information. WMO-No. 168. 6th ed. Geneva, 2008.

Referência de Implementação (Sistemas de Informação Geográfica):

ISAAKS, E. H.; SRIVASTAVA, R. M. An Introduction to Applied Geostatistics. Oxford University Press, 1989.

```python
# 1. Preparar os dados (usando seu df_anual limpo)
x = df_anual['longitude'].values
y = df_anual['latitude'].values
z = df_anual['Acumulado_Anual'].values

# 2. Definir os limites da grade (grid)
# Criamos uma margem de 0.05 graus (~5km) para cobrir as redondezas
grid_lon = np.linspace(x.min() - 0.05, x.max() + 0.05, 20)
grid_lat = np.linspace(y.min() - 0.05, y.max() + 0.05, 20)

# 3. Configurar e Executar a Krigagem Ordinária
# O modelo 'gaussian' ou 'spherical' costuma funcionar bem para chuvas
OK = OrdinaryKriging(
    x, y, z,
    variogram_model='gaussian',
    verbose=False,
    enable_plotting=False
)

# z_pred são os valores estimados, ss_pred é o erro (variância)
z_pred, ss_pred = OK.execute('grid', grid_lon, grid_lat)

# 4. Plotar o Mapa de Chuva Estimada
plt.figure(figsize=(10, 8))
# Invertemos a ordem de lat/lon no pcolormesh para alinhar com o mapa
plt.pcolormesh(grid_lon, grid_lat, z_pred, shading='auto', cmap='YlGnBu')
plt.colorbar(label='Precipitação Acumulada Estimada (mm)')

# Plotar as estações reais por cima para conferência
plt.scatter(x, y, c='red', marker='x', label='Estações Reais')

plt.title('Krigagem Ordinária: Distribuição Espacial de Chuva (SJC)', fontsize=14)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Listas para armazenar valores reais e previstos
valores_reais = z
valores_previstos = []

# Loop de Validação Cruzada Leave-One-Out
for i in range(len(z)):
    # Criar subconjuntos: tira o ponto 'i' do treino
    x_train = np.delete(x, i)
    y_train = np.delete(y, i)
    z_train = np.delete(z, i)

    # Treinar o modelo com os pontos restantes
    ok_cv = OrdinaryKriging(x_train, y_train, z_train,
                            variogram_model='spherical', verbose=False)

    # Prever o valor no ponto que foi retirado
    p, ss = ok_cv.execute('points', x[i], y[i])
    valores_previstos.append(p[0])

# Converter para array
valores_previstos = np.array(valores_previstos)

# Calcular Métricas
mae = mean_absolute_error(valores_reais, valores_previstos)
rmse = np.sqrt(mean_squared_error(valores_reais, valores_previstos))
r2 = r2_score(valores_reais, valores_previstos)

print(f"--- Métricas de Validação do Modelo ---")
print(f"MAE: {mae:.2f} mm")
print(f"RMSE: {rmse:.2f} mm")
print(f"R² Score: {r2:.4f}")

# Gráfico de Dispersão: Real vs Previsto
plt.figure(figsize=(7, 7))
plt.scatter(valores_reais, valores_previstos, color='blue', alpha=0.6)
plt.plot([valores_reais.min(), valores_reais.max()],
         [valores_reais.min(), valores_reais.max()], 'r--', lw=2)
plt.title('Validação Cruzada: Real vs. Previsto')
plt.xlabel('Chuva Real (mm)')
plt.ylabel('Chuva Prevista (Krigagem) (mm)')
plt.grid(True)
plt.show()
```

```python
# Criar um dataframe de conferência
df_validacao = pd.DataFrame({
    'Nome': df_anual['nome'],
    'Real': valores_reais,
    'Previsto': valores_previstos,
    'Erro_Absoluto': np.abs(valores_reais - valores_previstos)
})

# Mostrar as 5 estações onde o modelo mais errou
print(df_validacao.sort_values(by='Erro_Absoluto', ascending=False).head(5))
```

```python
def busca_refinada_krigagem(x, y, z):
    # Modelos mais estáveis primeiro
    modelos = ['spherical', 'exponential', 'gaussian']
    # Para poucos pontos, nlags deve ser baixo
    lags_test = [2, 4, 6]

    melhor_r2 = -np.inf
    melhor_config = None

    print("--- Iniciando Busca Refinada ---")

    for modelo in modelos:
        for nlags in lags_test:
            try:
                previsoes = []
                for i in range(len(z)):
                    x_train = np.delete(x, i)
                    y_train = np.delete(y, i)
                    z_train = np.delete(z, i)

                    # Ajuste: Adicionamos coordenadas exatas e removemos o excesso de parâmetros
                    ok = OrdinaryKriging(
                        x_train, y_train, z_train,
                        variogram_model=modelo,
                        nlags=nlags,
                        weight=True,
                        exact_values=False # Ajuda na estabilidade numérica
                    )

                    p, ss = ok.execute('points', x[i], y[i])
                    previsoes.append(p[0])

                previsoes = np.array(previsoes)
                r2 = r2_score(z, previsoes)
                rmse = np.sqrt(mean_squared_error(z, previsoes))

                print(f"Testado: {modelo} | Lags: {nlags} | R2: {r2:.4f}")

                if r2 > melhor_r2:
                    melhor_r2 = r2
                    melhor_config = {
                        'modelo': modelo,
                        'nlags': nlags,
                        'r2': r2,
                        'rmse': rmse,
                        'parametros_finais': ok.variogram_parameters
                    }
            except Exception as e:
                # Se falhar, agora ele avisa o porquê
                print(f"Erro no modelo {modelo} com {nlags} lags: {e}")
                continue

    return melhor_config

# Chame a função novamente
resultado_final = busca_refinada_krigagem(x, y, z)

if resultado_final:
    print("\n--- SUCESSO! MELHOR RESULTADO ---")
    print(resultado_final)
else:
    print("\n--- O modelo ainda não convergiu. Verifique se há pontos duplicados no seu DF. ---")
```

```python
# Configuração vencedora
modelo_vencedor = 'exponential'
lags_vencedor = 4

# Criar o modelo final com todos os dados
OK_final = OrdinaryKriging(
    x, y, z,
    variogram_model=modelo_vencedor,
    nlags=lags_vencedor,
    weight=True
)

# Agora executamos a grade 20x20
z_pred, ss_pred = OK_final.execute('grid', grid_lon, grid_lat)

# Acessar parâmetros com segurança nesta versão
try:
    # Em algumas versões os parâmetros ficam em variogram_model_parameters
    params = OK_final.variogram_model_parameters
    print(f"Parâmetros Otimizados:\nSill: {params[0]}\nRange: {params[1]}\nNugget: {params[2]}")
except:
    print("Nota: Parâmetros ajustados automaticamente pelo modelo.")

# Plotar o resultado final otimizado
plt.figure(figsize=(10, 8))
plt.pcolormesh(grid_lon, grid_lat, z_pred, shading='auto', cmap='Blues')
plt.colorbar(label='Precipitação (mm) - Modelo Exponencial')
plt.scatter(x, y, c='red', marker='x', label='Estações Reais')
plt.title(f'Mapa Final Otimizado (R²: 0.42)', fontsize=14)
plt.show()
```

# Dados População + Risco

```python
# A. Carregar o Shapefile de Inundação
risco_gdf = gpd.read_file('Dados/pddmap_so.shp').to_crs(epsg=4326)

# B. Função para extrair população do WorldPop de forma eficiente
def get_pop_data(lat_grid, lon_grid, path_worldpop):
    pop_values = []
    with rasterio.open(path_worldpop) as src:
        for lat, lon in zip(lat_grid, lon_grid):
            # Transforma coordenada geográfica em índice de pixel
            row, col = src.index(lon, lat)
            # Lê o valor do pixel (população estimada)
            val = src.read(1)[row, col]
            pop_values.append(val if val > 0 else 0)
    return np.array(pop_values)
```

```python
# C. Criar o DataFrame da Grade (20x20)
lats = np.linspace(-23.40, -22.70, 20)
lons = np.linspace(-46.05, -45.65, 20)
lon_mg, lat_mg = np.meshgrid(lons, lats)

df_grid = pd.DataFrame({
    'latitude': lat_mg.flatten(),
    'longitude': lon_mg.flatten()
})

# 1. Atribuir População
df_grid['populacao'] = get_pop_data(df_grid['latitude'], df_grid['longitude'], 'Dados/bra_pop_2025_CN_100m_R2025A_v1.tif')

# 2. Atribuir Risco (1 se estiver dentro do polígono do GeoSanja, 0 se não)
pontos_geometria = [Point(xy) for xy in zip(df_grid['longitude'], df_grid['latitude'])]
grid_gdf = gpd.GeoDataFrame(df_grid, geometry=pontos_geometria, crs="EPSG:4326")
df_grid['risco'] = grid_gdf.geometry.apply(lambda p: 1 if risco_gdf.contains(p).any() else 0)

# 3. Atribuir Chuva (Use o resultado da sua Krigagem Exponencial aqui)
# Supondo que você já tenha 'z_pred' da krigagem anterior
df_grid['chuva'] = z_pred.flatten() 

print("Grade integrada com sucesso!")
print(df_grid.head())
```

```python
# Normalização Min-Max para colocar tudo na escala de 0 a 1
def normalize(col):
    return (col - col.min()) / (col.max() - col.min()) if (col.max() - col.min()) != 0 else col

df_grid['chuva_n'] = normalize(df_grid['chuva'])
df_grid['pop_n'] = normalize(df_grid['populacao']) # Dado do WorldPop

# Cálculo do Peso Final (Sugestão de pesos para o artigo)
# Damos 40% de importância para População, 30% para Risco e 30% para Chuva
df_grid['W'] = (df_grid['pop_n'] * 0.4) + (df_grid['risco'] * 0.3) + (df_grid['chuva_n'] * 0.3)

# Salve agora para garantir o progresso
df_grid.to_csv('df_grid_final_com_pesos.csv', index=False)
```

# Comparação

```python
from scipy.spatial.distance import cdist

def avaliar_rede(df_estacoes, df_grid, raio_km=2.0):
    """
    Avalia a relevância de uma rede somando o peso W coberto 
    por um raio de influência ao redor de cada estação.
    """
    # Converter raio de km para graus aproximados (1 grau ~ 111km)
    raio_graus = raio_km / 111.0
    
    # Calcular distância de cada ponto do grid para cada estação da rede
    # matriz de distâncias (pontos_grid x estacoes)
    distancias = cdist(df_grid[['longitude', 'latitude']], df_estacoes[['longitude', 'latitude']])
    
    # Verificar quais pontos do grid estão dentro do raio de pelo menos uma estação
    coberto = np.any(distancias <= raio_graus, axis=1)
    
    # Somar o peso W total capturado
    valor_capturado = df_grid.loc[coberto, 'W'].sum()
    pop_coberta = df_grid.loc[coberto, 'populacao'].sum()
    pontos_risco_cobertos = df_grid.loc[coberto, 'risco'].sum()
    
    return valor_capturado, pop_coberta, pontos_risco_cobertos

# Criar DataFrame da rede atual para a função
df_rede_atual = pd.DataFrame({'longitude': x, 'latitude': y})

# Avaliar ambas
w_atual, pop_atual, risco_atual = avaliar_rede(df_rede_atual, df_grid)
w_otim, pop_otim, risco_otim = avaliar_rede(df_rede_otimizada, df_grid)

print("--- RESULTADO DA COMPARAÇÃO ---")
print(f"População 'protegida' (Rede Atual): {pop_atual:.0f}")
print(f"População 'protegida' (Rede Otimizada): {pop_otim:.0f}")
print(f"Melhoria na cobertura de população: {((pop_otim/pop_atual)-1)*100:.2f}%")
print(f"\nPontos de Risco monitorados (Rede Atual): {risco_atual}")
print(f"Pontos de Risco monitorados (Rede Otimizada): {risco_otim}")
```

```python
from scipy.spatial.distance import cdist

def avaliar_rede(df_estacoes, df_grid, raio_km=2.0):
    """
    Avalia a relevância de uma rede somando o peso W coberto 
    por um raio de influência ao redor de cada estação.
    """
    # Converter raio de km para graus aproximados (1 grau ~ 111km)
    raio_graus = raio_km / 111.0
    
    # Calcular distância de cada ponto do grid para cada estação da rede
    # matriz de distâncias (pontos_grid x estacoes)
    distancias = cdist(df_grid[['longitude', 'latitude']], df_estacoes[['longitude', 'latitude']])
    
    # Verificar quais pontos do grid estão dentro do raio de pelo menos uma estação
    coberto = np.any(distancias <= raio_graus, axis=1)
    
    # Somar o peso W total capturado
    valor_capturado = df_grid.loc[coberto, 'W'].sum()
    pop_coberta = df_grid.loc[coberto, 'populacao'].sum()
    pontos_risco_cobertos = df_grid.loc[coberto, 'risco'].sum()
    
    return valor_capturado, pop_coberta, pontos_risco_cobertos

# Criar DataFrame da rede atual para a função
df_rede_atual = pd.DataFrame({'longitude': x, 'latitude': y})

# Avaliar ambas
w_atual, pop_atual, risco_atual = avaliar_rede(df_rede_atual, df_grid)
w_otim, pop_otim, risco_otim = avaliar_rede(df_rede_equi, df_grid)

print("--- RESULTADO DA COMPARAÇÃO ---")
print(f"População 'protegida' (Rede Atual): {pop_atual:.0f}")
print(f"População 'protegida' (Rede Otimizada): {pop_otim:.0f}")
print(f"Melhoria na cobertura de população: {((pop_otim/pop_atual)-1)*100:.2f}%")
print(f"\nPontos de Risco monitorados (Rede Atual): {risco_atual}")
print(f"Pontos de Risco monitorados (Rede Otimizada): {risco_otim}")
```

```python
def avaliar_rede_completa(df_estacoes, df_grid, raio_km=2.0):
    # Raio de 2km convertido para graus (aprox.)
    raio_graus = raio_km / 111.0
    
    # Matriz de distâncias
    distancias = cdist(df_grid[['longitude', 'latitude']], df_estacoes[['longitude', 'latitude']])
    
    # Cobertura: Pontos do grid que estão no raio de pelo menos uma estação
    coberto = np.any(distancias <= raio_graus, axis=1)
    
    # Cálculos
    pop_total = df_grid.loc[coberto, 'populacao'].sum()
    risco_total = df_grid.loc[coberto, 'risco'].sum()
    
    # Chuva detectada: Média da chuva nos pontos cobertos pela rede
    # Isso mostra o "volume de chuva monitorado" pela rede
    chuva_media = df_grid.loc[coberto, 'chuva'].mean()
    
    return pop_total, risco_total, chuva_media

# Rodar a avaliação para os 3 cenários
# res_social veio do seu 1º teste (40/30/30)
# res_equi veio do seu 2º teste (33/33/33)
stats_atual = avaliar_rede_completa(df_rede_atual, df_grid)
stats_social = avaliar_rede_completa(df_rede_otimizada, df_grid)
stats_equi = avaliar_rede_completa(df_rede_equi, df_grid)

# Organizar para exibição
resultados = {
    "Métrica": ["População Protegida", "Pontos de Risco", "Chuva Média Detectada (mm)"],
    "Rede Atual": [stats_atual[0], stats_atual[1], stats_atual[2]],
    "Cenário Social (40/30/30)": [stats_social[0], stats_social[1], stats_social[2]],
    "Cenário Equilibrado (1/3)": [stats_equi[0], stats_equi[1], stats_equi[2]]
}

df_comparacao = pd.DataFrame(resultados)
print(df_comparacao)
```

# P-Medianas (abordagem heurística)

```python
def resolver_p_medianas(df_grid, n_p, peso_col='W'):
    """
    Heurística para encontrar p-medianas que minimizam a distância ponderada.
    """
    coords = df_grid[['latitude', 'longitude']].values
    pesos = df_grid[peso_col].values
    n_pontos = len(coords)
    
    # 1. Matriz de distâncias entre todos os pontos do grid
    dist_matrix = cdist(coords, coords)
    
    # 2. Iniciar com uma seleção aleatória ou baseada nos maiores pesos
    indices_atuais = df_grid.nlargest(n_p, peso_col).index.tolist()
    
    melhor_custo = np.sum(np.min(dist_matrix[:, indices_atuais], axis=1) * pesos)
    
    # 3. Busca local simples (Troca de pontos)
    for _ in range(5): # Rodar algumas iterações de refinamento
        for i in range(n_p):
            for j in range(n_pontos):
                if j not in indices_atuais:
                    novos_indices = indices_atuais.copy()
                    novos_indices[i] = j
                    
                    novo_custo = np.sum(np.min(dist_matrix[:, novos_indices], axis=1) * pesos)
                    
                    if novo_custo < melhor_custo:
                        melhor_custo = novo_custo
                        indices_atuais = novos_indices
                        
    return df_grid.iloc[indices_atuais]

# Executar para o cenário Social (Peso W)
df_p_medianas = resolver_p_medianas(df_grid, n_estacoes, peso_col='W')

print("P-Medianas calculadas!")
```

# Abordatem Hibrida

```python
def resolver_hibrido_kmeans_pmed(df_grid, n_p, peso_col='W'):
    # 1. Fase K-means: Definir as regiões (clusters)
    kmeans = KMeans(n_clusters=n_p, random_state=42, n_init=10)
    # Criamos os clusters baseados na localização, ponderados pelo peso
    clusters = kmeans.fit_predict(df_grid[['longitude', 'latitude']], sample_weight=df_grid[peso_col])
    df_grid['cluster_id'] = clusters
    
    estacoes_hibridas = []
    
    # 2. Fase P-medianas: Otimizar local dentro de cada cluster
    for i in range(n_p):
        cluster_data = df_grid[df_grid['cluster_id'] == i]
        
        if not cluster_data.empty:
            # Dentro do cluster, qual ponto minimiza a distância para os outros pontos do mesmo cluster?
            coords_cluster = cluster_data[['latitude', 'longitude']].values
            pesos_cluster = cluster_data[peso_col].values
            
            # Matriz de distâncias interna do cluster
            dist_matrix = cdist(coords_cluster, coords_cluster)
            
            # O ponto 'mediana' do cluster é aquele onde a soma das distâncias * peso é mínima
            soma_distancias_ponderadas = np.sum(dist_matrix * pesos_cluster, axis=1)
            melhor_ponto_idx = np.argmin(soma_distancias_ponderadas)
            
            estacoes_hibridas.append(cluster_data.iloc[melhor_ponto_idx])
            
    return pd.DataFrame(estacoes_hibridas)

# Executar a nova abordagem
df_rede_hibrida = resolver_hibrido_kmeans_pmed(df_grid, n_estacoes)

# Avaliar
stats_hibrida = avaliar_rede_completa(df_rede_hibrida, df_grid, raio_km=2.0)
```

```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import cdist
from sklearn.cluster import KMeans

# 1. Ajuste da Função de Avaliação (Contagem Única/Exclusiva)
def avaliar_rede_exclusiva(df_estacoes, df_grid, raio_km=2.0):
    raio_graus = raio_km / 111.0
    coords_grid = df_grid[['latitude', 'longitude']].values
    coords_estacoes = df_estacoes[['latitude', 'longitude']].values
    
    # Matriz de distâncias (Pontos Grid x Estações)
    distancias = cdist(coords_grid, coords_estacoes)
    
    # Mascara: True se o ponto está coberto por pelo menos UMA estação
    coberto = np.any(distancias <= raio_graus, axis=1)
    
    pop_total = df_grid.loc[coberto, 'populacao'].sum()
    risco_total = df_grid.loc[coberto, 'risco'].sum()
    chuva_media = df_grid.loc[coberto, 'chuva'].mean()
    
    return pop_total, risco_total, chuva_media

# 2. Loop de Experimentos
# Definindo as configurações conforme seu pedido
configuracoes = [
    {"nome": "Balanceado (1/3)", "pop": 0.333, "risco": 0.333, "chuva": 0.334},
    {"nome": "Foco População (50%)", "pop": 0.50, "risco": 0.25, "chuva": 0.25},
    {"nome": "Foco Risco (50%)", "pop": 0.25, "risco": 0.50, "chuva": 0.25},
    {"nome": "Foco Chuva (50%)", "pop": 0.25, "risco": 0.25, "chuva": 0.50}
]
fatores_risco = [1, 50, 100]

# Armazenar estatísticas da rede real para comparação
# Certifique-se que df_rede_atual (x, y) já existe no seu notebook
stats_real = avaliar_rede_exclusiva(df_rede_atual, df_grid)

for conf in configuracoes:
    for f_risco in fatores_risco:
        # Criar Peso W para este cenário
        W = (df_grid['pop_n'] * conf['pop']) + \
            ((df_grid['risco'] * f_risco) * conf['risco']) + \
            (df_grid['chuva_n'] * conf['chuva'])
        df_grid['W_exp'] = W

        # --- Executar Otimizadores ---
        # A) K-Means
        km = KMeans(n_clusters=n_estacoes, random_state=42, n_init=10)
        km.fit(df_grid[['longitude', 'latitude']], sample_weight=df_grid['W_exp'])
        df_km = pd.DataFrame(km.cluster_centers_, columns=['longitude', 'latitude'])

        # B) P-Medians (Seleção direta baseada em peso para velocidade)
        df_pmed = df_grid.nlargest(n_estacoes, 'W_exp')[['latitude', 'longitude']]

        # C) Híbrido
        df_grid['cluster_id'] = km.labels_
        hibrido_coords = []
        for i in range(n_estacoes):
            cluster_data = df_grid[df_grid['cluster_id'] == i]
            if not cluster_data.empty:
                c_coords = cluster_data[['latitude', 'longitude']].values
                c_pesos = cluster_data['W_exp'].values
                # Mediana do cluster
                d_mat = cdist(c_coords, c_coords)
                idx = np.argmin(np.sum(d_mat * c_pesos, axis=1))
                hibrido_coords.append(cluster_data.iloc[idx][['latitude', 'longitude']])
        df_hib = pd.DataFrame(hibrido_coords)

        # --- Avaliação e Tabela ---
        res_km = avaliar_rede_exclusiva(df_km, df_grid)
        res_pm = avaliar_rede_exclusiva(df_pmed, df_grid)
        res_hb = avaliar_rede_exclusiva(df_hib, df_grid)

        data = [
            ["Rede Vigente", stats_real[0], 0, stats_real[1], 0, stats_real[2], 0],
            ["K-Means", res_km[0], ((res_km[0]/stats_real[0])-1)*100, res_km[1], ((res_km[1]/stats_real[1])-1)*100, res_km[2], ((res_km[2]/stats_real[2])-1)*100],
            ["P-Medians", res_pm[0], ((res_pm[0]/stats_real[0])-1)*100, res_pm[1], ((res_pm[1]/stats_real[1])-1)*100, res_pm[2], ((res_pm[2]/stats_real[2])-1)*100],
            ["Híbrido", res_hb[0], ((res_hb[0]/stats_real[0])-1)*100, res_hb[1], ((res_hb[1]/stats_real[1])-1)*100, res_hb[2], ((res_hb[2]/stats_real[2])-1)*100]
        ]

        df_res = pd.DataFrame(data, columns=["Método", "População", "Ganho Pop %", "Risco", "Ganho Risco %", "Chuva", "Ganho Chuva %"])
        
        print(f"\n" + "="*90)
        print(f"EXPERIMENTO: {conf['nome']} | FATOR RISCO: {f_risco}")
        print("="*90)
        # ESPECIFICADOR CORRIGIDO: .2f
        print(df_res.to_string(index=False, float_format=lambda x: "{:.2f}".format(x)))
```
